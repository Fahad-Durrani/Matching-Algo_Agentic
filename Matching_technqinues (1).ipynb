```python
import pandas as pd
from datetime import datetime

# Helper function to calculate experience duration in years and months
def calculate_total_experience(start_date: str, end_date: str) -> dict:
    # Check if start_date is just a year (e.g., "2017")
    if len(start_date) == 4 and start_date.isdigit():
        start_date = f"January {start_date}"  # Assume January of that year

    # Try parsing both full and abbreviated month names
    try:
        start = datetime.strptime(start_date, '%B %Y')  # Full month name (e.g., "January 2022")
    except ValueError:
        try:
            start = datetime.strptime(start_date, '%b %Y')  # Abbreviated month name (e.g., "Jan 2022")
        except ValueError:
            start = datetime.strptime(start_date, '%Y')  # Just the year (e.g., "2017")

    # Handle "Present" or "Current" as the end date
    if end_date.lower() in ['present', 'current']:
        end = datetime.now()
    else:
        # Check if end_date is just a year (e.g., "2023")
        if len(end_date) == 4 and end_date.isdigit():
            end_date = f"January {end_date}"  # Assume January of that year

        # Try parsing both full and abbreviated month names for the end date
        try:
            end = datetime.strptime(end_date, '%B %Y')  # Full month name (e.g., "January 2022")
        except ValueError:
            try:
                end = datetime.strptime(end_date, '%b %Y')  # Abbreviated month name (e.g., "Jan 2022")
            except ValueError:
                end = datetime.strptime(end_date, '%Y')  # Just the year (e.g., "2023")

    # Calculate difference in years and months
    delta_years = end.year - start.year
    delta_months = end.month - start.month

    # Adjust for negative months (if the current month is less than the start month)
    if delta_months < 0:
        delta_years -= 1
        delta_months += 12

    return {'year': delta_years, 'month': delta_months}

# Helper function to format start and end dates as "(mm-yy)"
def format_start_end_dates(start, end):
    # Check if start is just a year (e.g., "2017")
    if len(start) == 4 and start.isdigit():
        start = f"January {start}"  # Assume January of that year

    # Try parsing both full and abbreviated month names
    try:
        start_date = datetime.strptime(start, '%B %Y')
    except ValueError:
        try:
            start_date = datetime.strptime(start, '%b %Y')
        except ValueError:
            start_date = datetime.strptime(start, '%Y')

    if end.lower() in ['present', 'current']:  # Handle both 'Present' and 'Current'
        end_date = datetime.now()
    else:
        # Check if end is just a year (e.g., "2023")
        if len(end) == 4 and end.isdigit():
            end = f"January {end}"  # Assume January of that year

        # Try parsing both full and abbreviated month names for the end date
        try:
            end_date = datetime.strptime(end, '%B %Y')
        except ValueError:
            try:
                end_date = datetime.strptime(end, '%b %Y')
            except ValueError:
                end_date = datetime.strptime(end, '%Y')

    # Format the start and end dates as "mm-yy"
    return f"{start_date.month}-{str(start_date.year)[-2:]}", f"{end_date.month}-{str(end_date.year)[-2:]}"



# Process each resume
all_candidates = []

for file_name, employment_history in zip(file_names, resumes):
    job_titles = [job['job_title'] for job in employment_history]
    job_responsibilities = [job['job_responsibilities'] for job in employment_history]

    experience_start_end_dates = []
    experience_durations = []
    total_years = 0
    total_months = 0

    for job in employment_history:
        start_date = job['start_date']
        end_date = job['end_date']

        # Calculate experience for the job
        duration = calculate_total_experience(start_date, end_date)
        total_years += duration['year']
        total_months += duration['month']

        # Format start and end dates
        start_date_formatted, end_date_formatted = format_start_end_dates(start_date, end_date)
        experience_start_end_dates.append(f"({start_date_formatted}, {end_date_formatted})")
        experience_durations.append(f"{duration['year']}.{duration['month']}")

    # Add extra months to total years if applicable
    total_years += total_months // 12
    total_months = total_months % 12

    # Final total experience in "years.months" format
    total_experience = f"{total_years}.{total_months}"

    # Create the candidate info dictionary
    candidate_info = {
        "candidate": file_name,
        "job_title": job_titles,
        "total_experience": total_experience,  # e.g., "4.9"
        "experiences": len(employment_history),
        "experience_start_end_dates": experience_start_end_dates,  # e.g., [(7-22, 11-24), ...]
        "experience_durations": experience_durations,  # e.g., ["2.4", "2.5"]
        "job_responsibilities": job_responsibilities
    }

    all_candidates.append(candidate_info)

# Convert to pandas DataFrame
df = pd.DataFrame(all_candidates)

# Display the DataFrame
print(df)
```
