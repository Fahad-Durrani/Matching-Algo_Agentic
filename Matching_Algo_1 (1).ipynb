# **Poppler**

```python
#popler for PDF to text
def pdftotext(path):
    """
    Generate a text rendering of a PDF file in the form of a list of lines.
    """
    #args = ['pdftotext', '-layout', path, '-']
    args = ['pdftotext', path, '-']
    cp = sp.run(
      args, stdout=sp.PIPE, stderr=sp.DEVNULL,
      check=True, text=True
    )
    return cp.stdout
```

```python
def analyze_read(file_stream):
    document_analysis_client = DocumentAnalysisClient(
        endpoint=endpoint, credential=credential
    )

    poller = document_analysis_client.begin_analyze_document(
        "prebuilt-read", file_stream
    )
    result = poller.result()

    text_content = []
    for page in result.pages:
        for line in page.lines:
            text_content.append(line.content)

    return "\n".join(text_content)
```

```python
@retry(Exception, tries=3, timeout_secs=10)
def call_gpt_cv(query):
  client = OpenAI(api_key=api_key)
  completion = client.chat.completions.create(
      model="gpt-4o-2024-08-06",
      messages=[
          {"role": "system",
            "content": (
            "You are a helpful information extraction system. You will be provided with a resume or CV text and your task is to extract the useful information such as name, address, email, skills etc. in JSON format."
            "Only use the provided resume or CV text to extract the useful information except for industry the resume belongs to."
            f"Additionally, assign the 'career_level' field based on the following criteria only:{career_level}"
            ),
            },
          {"role": "user", "content": query}
      ],
      temperature=0,
      # max_tokens=4096,
      seed = 1,
      tools=functions,
      tool_choice='required'
  )

  return completion
```

```python
#---RESUME PARSING PIPELINE
def resume_parsing(pdf_files_resumes):
  for i in range(len(pdf_files_resumes)):
    resume = pdftotext(pdf_files_resumes[i])
    response = call_gpt_cv(resume)
    resume_dict=response.choices[0].message.tool_calls[0].function.arguments
    resume_dict = json.loads(resume_dict)
    file_name = f'resume_{1 + i}.json'
    base_folder = folder_path='/content/drive/My Drive/Matching-Algo-Takafo/Resumes/Machine Learning/json'
    json_file_path = os.path.join(base_folder, file_name)
    # Save to Google Drive
    with open(json_file_path, 'w') as json_file:
      json.dump(resume_dict, json_file, indent=4)
  return None
```
